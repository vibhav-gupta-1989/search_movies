# Load the movie release years, titles and plots

```python
import pandas as pd

df = pd.read_csv('wiki_movie_plots_deduped.csv', usecols=['Release Year', 'Title', 'Plot'])
print(len(df))
print(df['Plot'][0])
```

# Store the movie plots in an array

```python
movie_plots_array = df['Plot'].to_numpy()
print(movie_plots_array[1])
```

# Input search query

```python
query = "A young boy discovers he has magical powers and attends a school for wizards."
```

# Load the SBERT model

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
```

# Embed the movie plots

```python
embeddings = model.encode(movie_plots_array, convert_to_tensor=True)
print(embeddings.shape)
```

# Embed the input query

```python
query_embedding = model.encode(query, convert_to_tensor=True)
```

# Compute similarities

```python
similarities = model.similarity(query_embedding, embeddings)
len(similarities[0])
```

# Return 5 most similar movie names

```python
idx = similarities[0].argsort(descending=True)[:5]
print("Top 5 most similar movies:")
for i in idx:
    i = i.item()  # Convert tensor to integer
    print(f"Release Year: {df['Release Year'][i]}")
    print(f"Title: {df['Title'][i]}")
    print(f"Similarity Score: {similarities[0][i]:.4f}")
    print("-" * 50)
```
